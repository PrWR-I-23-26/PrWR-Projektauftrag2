```python
import pandas as pd
import plotly.express as px

# Daten hole
bevölkerung_df = pd.read_csv("KTZH_00001521_00002899.csv")
beschäftigte_df = pd.read_csv("KTZH_00001823_00003360.csv")

# Gesamtbevölkerig pro Jahr summiere
bevölkerung_sum = bevölkerung_df.groupby("jahr")["anzahl"].sum().reset_index()
bevölkerung_sum.rename(columns={"anzahl": "gesamtbevölkerung"}, inplace=True)

# Beschäftigtendate pro Jahr zusammenführe mit Mittelwert
beschäftigte_avg = beschäftigte_df.groupby("jahr")["beschaeftigte"].mean().reset_index()
beschäftigte_avg.rename(columns={"beschaeftigte": "durchschnittliche_beschaeftigte"}, inplace=True)

# Beschäftigti nach Geschlecht 
beschäftigte_gender = beschäftigte_df.groupby(["jahr"])[["beschaeftigte_frauen", "beschaeftigte_maenner"]].mean().reset_index()

# Datenquelle verknüpfe
merged_df = pd.merge(bevölkerung_sum, beschäftigte_avg, on="jahr", how="inner")
merged_df = pd.merge(merged_df, beschäftigte_gender, on="jahr", how="inner")

# Interaktivi Visualisierig mit Plotly
fig = px.line(merged_df, x="jahr", y=["gesamtbevölkerung", "durchschnittliche_beschaeftigte", "beschaeftigte_frauen", "beschaeftigte_maenner"],
              labels={"value": "Anzahl", "variable": "Kategorie"},
              title="Entwicklung der Bevölkerung und Beschäftigten in Winterthur",
              markers=True)
fig.update_layout(hovermode="x unified")

# Dropdown für verschideni Kategorie
fig.update_layout(
    updatemenus=[
        {
            "buttons": [
                {"label": "Alle", "method": "update", "args": [{"visible": [True, True, True, True]}]},
                {"label": "Bevölkerung & Beschäftigte", "method": "update", "args": [{"visible": [True, True, False, False]}]},
                {"label": "Bevölkerung", "method": "update", "args": [{"visible": [True, False, False, False]}]},
                {"label": "Gesamtbeschäftigte", "method": "update", "args": [{"visible": [False, True, False, False]}]},
                {"label": "Frauen", "method": "update", "args": [{"visible": [False, False, True, False]}]},
                {"label": "Männer", "method": "update", "args": [{"visible": [False, False, False, True]}]},
            ],
            "direction": "down",
            "showactive": True,
        }
    ]
)

fig.show()
```


Entwicklung der Bevölkerung und Beschäftigten in Winterthur (2018–2023)

Die Daten zeigen ein kontinuierliches Bevölkerungswachstum sowie eine teils schwankende, aber insgesamt positive Entwicklung bei den Beschäftigtenzahlen.

Bevölkerungswachstum:
Von 114.182 (2018) auf 121.652 (2023) gestiegen.
Trotz der COVID-19-Pandemie setzte sich das Wachstum fort, was auf Winterthur als attraktiven Wohnort hinweist.
Besonders nach 2021 war ein starker Anstieg zu beobachten, möglicherweise durch wirtschaftliche Erholung und Zuwanderung.


Beschäftigungsentwicklung:
- 2018–2019: Steigerung von 73.200 auf 74.625, wirtschaftlich stabile Phase.
- 2020: Leichter Rückgang (74.500) durch die Pandemie, Erholung ab 2021.
- 2023: Sprung auf 77.000, wohl durch wirtschaftliche Erholung und neue Arbeitsplätze.


Beschäftigung nach Geschlecht:
- Frauen: Stetiger Anstieg von 35.350 (2018) auf 37.700 (2023), trotz kurzer Stagnation 2020.
- Männer: Schwankungen, aber letztlich Anstieg von 37.850 (2018) auf 39.325 (2023), nach Einbruch 2020–2022.


Fazit:
Winterthur wuchs trotz Krisenzeiten weiter. Die Pandemie bremste den Arbeitsmarkt nur kurzfristig, und 2023 deutet auf eine starke wirtschaftliche Erholung hin.
